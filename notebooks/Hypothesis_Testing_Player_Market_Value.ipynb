```python
import pandas as pd
from scipy.stats import pearsonr

df = pd.read_excel("yigitdsa210data.xlsx")

def convert_market_value(x):
    if pd.isna(x):
        return None

    x = str(x).strip()
    x = x.replace("€", "").lower()

    if x.endswith("m"):
        return float(x[:-1]) * 1000000
    elif x.endswith("k"):
        return float(x[:-1]) * 1000
    else:
        return float(x)

df["Market Value Numeric"] = df["Market Value"].apply(convert_market_value)
```

## Hypothesis 0

H0: There is no significant relationship between G+A (Goals + Assists) and player market value.

H1: There is a significant relationship between G+A (Goals + Assists) and player market value.

```python
df["G+A"] = pd.to_numeric(df["G+A"], errors="coerce")

temp_df_ga = df[["G+A", "Market Value Numeric"]].dropna()

correlation, p_value = pearsonr(
    temp_df_ga["G+A"],
    temp_df_ga["Market Value Numeric"]
)

print("Correlation:", correlation)
print("P-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```

### Interpretation

Since the p-value is smaller than 0.05, the null hypothesis is rejected.

This indicates that there is a statistically significant relationship between attacking performance (G+A) and player market value.

The positive correlation suggests that players with higher goals and assists generally tend to have higher market values.

## Hypothesis 1

H0: Age has no significant relationship with player market value.

H1: Age has a significant relationship with player market value.

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

temp_df_age = df[["Age", "Market Value Numeric"]].dropna()

correlation_age, p_value_age = pearsonr(
    temp_df_age["Age"],
    temp_df_age["Market Value Numeric"]
)

print("Correlation:", correlation_age)
print("P-value:", p_value_age)

alpha = 0.05

if p_value_age < alpha:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```

### Interpretation

Since the p-value is greater than 0.05, the null hypothesis cannot be rejected.

This suggests that age does not have a statistically significant relationship with player market value in this dataset.

The weak negative correlation also indicates that age alone is not a strong predictor of market value.

## Hypothesis 2

H0: Minutes played have no significant relationship with player market value.

H1: Minutes played have a significant relationship with player market value.

```python
df["Min"] = pd.to_numeric(df["Min"], errors="coerce")

temp_df = df[["Min", "Market Value Numeric"]].dropna()

correlation_min, p_value_min = pearsonr(
    temp_df["Min"],
    temp_df["Market Value Numeric"]
)

print("Correlation:", correlation_min)
print("P-value:", p_value_min)

alpha = 0.05

if p_value_min < alpha:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```

### Interpretation

Since the p-value is smaller than 0.05, the null hypothesis is rejected.

This indicates that minutes played have a statistically significant relationship with player market value.

The negative correlation suggests that players with higher playing time do not necessarily have higher market values in this dataset.

# Overall Conclusion

The hypothesis tests and machine learning models show that attacking performance variables, especially goals and assists (G+A), have the strongest relationship with player market value.

Age alone was not found to be statistically significant, while minutes played showed a weak negative relationship with market value.

Overall, player market value appears to be influenced more by attacking contribution than by simple playing time or age.
