```python
import pandas as pd

df = pd.read_excel("/content/sample_data/yigitdsa210data.xlsx")
df.head()
```

# DSA210 Project - Yiğit

In this project, I analyze how different factors relate to player market value.

```python
def clean_value(x):
    if isinstance(x, str):
        x = x.replace("€", "").replace("m", "")
        return float(x) * 1_000_000
    return x

df["Market Value"] = df["Market Value"].apply(clean_value)
```

```python
from scipy.stats import pearsonr

corr, p_value = pearsonr(df["Age"], df["Market Value"])

print("Correlation:", corr)
print("P-value:", p_value)
```

The correlation between age and market value is low.

This means age alone does not strongly affect market value.

```python
df = df.dropna(subset=["Age", "Market Value"])
```

```python
from scipy.stats import pearsonr

corr, p_value = pearsonr(df["Age"], df["Market Value"])

print("Correlation:", corr)
print("P-value:", p_value)
```

The correlation between age and market value is very low and slightly negative.

This means age does not have a strong effect on market value.

The p-value is high, so the result is not statistically significant.

Therefore, we cannot reject the null hypothesis.

```python
import matplotlib.pyplot as plt

plt.scatter(df["Age"], df["Market Value"])
plt.xlabel("Age")
plt.ylabel("Market Value")
plt.title("Age vs Market Value")
plt.show()
```
